## What You Are Asked To Do

**Does temperature explain daily deaths in Chicago, and what would a
model that says so actually be claiming?**

That is the question. Answer it by running the four-step loop end to
end: **specify** a generative model, **simulate** from it, **fit** it,
and **test** whether you should believe the fit. All four steps are
required.

**How you organize the writeup is up to you.** There are no numbered
problems to work through in order. You are writing a short report that
makes one argument and shows the evidence behind it, and deciding what
to lead with, how much space each piece deserves, and what belongs in a
figure rather than a paragraph is part of the work.

### What your report must contain

Your report is graded on whether these are present and convincing, **not
on whether they appear in this order or under these headings.**

- **A specification** *(10 points — Specifying a Model)*. In
  mathematical notation, with the distribution justified from a
  constraint you are willing to state about the process, and **every**
  assumption named.
- **Evidence you checked the specification before trusting it** *(8
  points — Simulating From a Model)*. Simulate from it, with parameters
  you chose, and compare against the real data.
- **A fit** *(10 points — Fitting and Recovery)*. By a method you can
  justify, with each parameter interpreted in the context of deaths and
  degrees.
- **A test of whether your fitting procedure recovers parameters you
  know** *(10 points — Fitting and Recovery)*. Simulate from the fit,
  refit, and compare against what you put in.
- **Honest uncertainty** *(7 points — Interpretation)*. On the one
  quantity that matters most for the question, with a statement of what
  that interval is an interval *on*.
- **An argument about what your model cannot do** *(5 points —
  Interpretation)*. What it assumes that you do not believe, where it
  does worst, and what you would change first.

**50 points total.** The point value beside each element is what it is
worth; the [standard
rubrics](https://envdata.viveks.me/spring2027/policies/rubric.html)
describe how each kind of element is marked.

### Form

No more than **four pages** excluding figures and references. You do not
need to include code, but your notebook or scripts must be in your
repository. Raw program output is not an answer to anything — write in
sentences.

> **The assumptions are the assignment**
>
> A specification with no assumptions stated earns nothing for that
> element, whatever else the report does. This is the single thing the
> course most wants you to stop doing.

> **Simulate before you fit**
>
> Simulating from a specification you have not yet fitted is the
> cheapest check in the loop and the one most often skipped. Fitting
> only finds the best parameters *within* a model you have already
> committed to — if the specification cannot produce data resembling
> these observations, a good fit will not tell you so.

> **Report what you find, not what you hoped**
>
> A recovery test that fails, honestly reported and diagnosed, earns
> full credit. A recovery test that fails and is not noticed earns none.

> **You are expected to find problems**
>
> This project is built so that a careful specification still leaves
> things on the table, and several of them are things we return to later
> in the semester. **Finding them yourself and saying so is worth more
> than a model that appears to have no flaws.**

> **BEE 5850**
>
> Students in BEE 5850 must **also** fit the model with a Bayesian
> treatment: state your priors and justify each one, sample the
> posterior, and report posterior summaries alongside the maximum
> likelihood estimates. Say where the two agree, where they do not, and
> what the prior is doing at this sample size.
