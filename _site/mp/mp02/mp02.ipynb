## What You Are Asked To Do

**What is the 100-year flood on this river, and how much of your answer
comes from the river rather than from your choices?**

Answer it by fitting a **GEV** to annual maxima and a **GPD** to peaks
over a threshold, each in a **stationary** and a **non-stationary** form
— four models — and comparing them on evidence.

**How you organize the writeup is up to you.** There are no numbered
problems. You are writing a short report that makes one argument and
shows the evidence behind it.

### What your report must contain

Your report is graded on whether these are present and convincing, **not
on whether they appear in this order or under these headings.**

- **A defended decision about which record you are analysing** *(10
  points — Specifying a Model)*. Which years you used, what you did
  about the absent and partial ones, and why. Say what a reader would

- **Four fitted models** *(10 points — Fitting and Recovery)*. GEV and
  GPD, each stationary and non-stationary. State your threshold and
